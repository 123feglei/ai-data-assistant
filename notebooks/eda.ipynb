# 智能数据分析助手 · D2 数据探索（EDA）

数据源：巴西电商 Olist（8 张业务表 + geolocation 辅助表），数据库：`data/ecommerce.db`。

目的：① 验证数据质量与口径；② 为 `data/schema.md`（字段中文说明）与 `data/metrics.md`（指标清单）提供依据。

```python
import sqlite3
from pathlib import Path

import pandas as pd
import plotly.express as px

# 定位项目根目录（本 notebook 位于 notebooks/ 下，兼容从根目录或 notebooks/ 启动）
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'data' / 'ecommerce.db').exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
DB_PATH = PROJECT_ROOT / 'data' / 'ecommerce.db'

conn = sqlite3.connect(DB_PATH)

def q(sql: str) -> pd.DataFrame:
    '''执行 SQL 返回 DataFrame'''
    return pd.read_sql_query(sql, conn)

print('数据库:', DB_PATH)
```

## 1. 数据概览：9 张表

```python
tables = q("SELECT name FROM sqlite_master WHERE type='table' ORDER BY name")
rows = []
for t in tables['name']:
    n = q(f'SELECT COUNT(*) n FROM "{t}"').iloc[0, 0]
    cols = q(f'SELECT * FROM "{t}" LIMIT 0').columns.tolist()
    rows.append({'表名': t, '行数': int(n), '列数': len(cols), '字段': ', '.join(cols)})
overview = pd.DataFrame(rows)
overview
```

## 2. 订单状态分布（数据质量）

```python
status_df = q("SELECT order_status, COUNT(*) n FROM orders "
              "GROUP BY order_status ORDER BY n DESC")
status_df['占比%'] = (status_df['n'] / status_df['n'].sum() * 100).round(2)
px.bar(status_df, x='order_status', y='n', text='n',
       title='订单状态分布', labels={'order_status': '状态', 'n': '订单数'})
```

## 3. 整体业务指标（GMV / 客单价 / 复购率 / 配送）

```python
# 在 notebook 内重新计算整体指标（供后续单元使用，与构建脚本同一口径）
orders = int(q('SELECT COUNT(DISTINCT order_id) n FROM orders').iloc[0, 0])
gmv = float(q('SELECT ROUND(SUM(price),2) v FROM order_items').iloc[0, 0])
aov = float(q('SELECT ROUND(SUM(price)*1.0/COUNT(DISTINCT order_id),2) v '
              'FROM order_items').iloc[0, 0])
repurchase = float(q("""SELECT ROUND(
                       SUM(CASE WHEN buy_times>=2 THEN 1 ELSE 0 END)*100.0/COUNT(*),2) v
                   FROM (SELECT c.customer_unique_id, COUNT(DISTINCT o.order_id) buy_times
                         FROM customers c JOIN orders o ON c.customer_id=o.customer_id
                         WHERE o.order_status NOT IN ('canceled','unavailable')
                         GROUP BY c.customer_unique_id)""").iloc[0, 0])
avg_delivery = float(q("""SELECT ROUND(AVG(julianday(order_delivered_customer_date)
                          - julianday(order_purchase_timestamp)),1) v
                      FROM orders
                      WHERE order_status='delivered' AND order_delivered_customer_date IS NOT NULL""").iloc[0, 0])
date_range = q('SELECT MIN(order_purchase_timestamp) lo, '
               'MAX(order_purchase_timestamp) hi FROM orders').iloc[0]
print(f'总订单数: {orders:,}')
print(f'GMV: {gmv:,.2f}')
print(f'客单价: {aov}')
print(f'复购率: {repurchase}%')
print(f'平均配送时长: {avg_delivery} 天')
print(f'数据范围: {date_range["lo"]} ~ {date_range["hi"]}')
```

## 4. 各州订单分布（地理）

```python
state_df = q("""SELECT c.customer_state s, COUNT(DISTINCT o.order_id) n
               FROM orders o JOIN customers c ON o.customer_id=c.customer_id
               GROUP BY s ORDER BY n DESC""")
px.bar(state_df.head(10), x='s', y='n', text='n',
       title='订单量 TOP10 州', labels={'s': '州', 'n': '订单数'})
```

## 5. 品类销量 TOP10

```python
cat_df = q("""SELECT t.product_category_name_english cat, SUM(oi.price) v
             FROM order_items oi
             JOIN products p ON oi.product_id=p.product_id
             JOIN product_category_name_translation t
                  ON p.product_category_name=t.product_category_name
             GROUP BY cat ORDER BY v DESC LIMIT 10""")
px.bar(cat_df, x='cat', y='v', text='v',
       title='品类销量 TOP10', labels={'cat': '品类', 'v': '销量(GMV)'})
```

## 6. 支付方式分布

```python
pay_df = q("""SELECT payment_type s, COUNT(DISTINCT order_id) n,
                 ROUND(SUM(payment_value),2) v
             FROM order_payments GROUP BY s ORDER BY n DESC""")
px.pie(pay_df, names='s', values='n', title='支付方式订单占比',
       labels={'s': '支付方式', 'n': '订单数'})
```

## 7. 月度订单量趋势（时间序列）

```python
month_df = q("""SELECT strftime('%Y-%m', order_purchase_timestamp) m,
                  COUNT(DISTINCT order_id) n
              FROM orders WHERE order_status != 'canceled'
              GROUP BY m ORDER BY m""")
px.line(month_df, x='m', y='n', markers=True,
       title='月度订单量趋势', labels={'m': '月份', 'n': '订单数'})
```

## 8. 评分与配送时长分布

```python
score_df = q("SELECT review_score, COUNT(*) n FROM order_reviews "
             "GROUP BY review_score ORDER BY review_score")
px.bar(score_df, x='review_score', y='n', text='n',
       title='订单评分分布', labels={'review_score': '评分', 'n': '数量'})
```

```python
deliv_df = q("""SELECT CAST(julianday(order_delivered_customer_date)
                    - julianday(order_purchase_timestamp) AS INT) days
              FROM orders
              WHERE order_status='delivered' AND order_delivered_customer_date IS NOT NULL""")
px.histogram(deliv_df, x='days', nbins=30, title='配送时长分布（天）',
            labels={'days': '配送天数'})
```

## 9. 数据质量：空值检查

```python
quality = []
for t in tables['name']:
    for col in q(f'SELECT * FROM "{t}" LIMIT 0').columns:
        nulls = q(f'SELECT COUNT(*) FROM "{t}" WHERE "{col}" IS NULL OR "{col}" = \'\'').iloc[0, 0]
        if nulls > 0:
            quality.append({'表': t, '字段': col, '空值数': int(nulls)})
quality_df = pd.DataFrame(quality).sort_values('空值数', ascending=False)
print(f'发现 {len(quality_df)} 个含空值的字段：')
quality_df.head(20)
```
